```python
def calculate_cost(state):
    """Calculates cost as the number of attacking pairs (diagonal conflicts)."""
    attacks = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            if abs(state[i] - state[j]) == abs(i - j):
                attacks += 1
    return attacks

def get_best_successor(state):
    """Generates all neighbors by swapping columns and returns the lowest-cost one."""
    best_neighbor = None
    min_cost = float('inf')
    best_swap = None
    all_successors = []
    n = len(state)

    for i in range(n):
        for j in range(i + 1, n):
            neighbor = list(state)
            neighbor[i], neighbor[j] = neighbor[j], neighbor[i]
            neighbor_tuple = tuple(neighbor)
            cost = calculate_cost(neighbor_tuple)
            all_successors.append((neighbor_tuple, cost, (i, j)))

            if cost < min_cost:
                min_cost = cost
                best_neighbor = neighbor_tuple
                best_swap = (i, j)
            elif cost == min_cost and best_swap is not None:
                if (i, j) < best_swap:
                    best_neighbor = neighbor_tuple
                    best_swap = (i, j)

    return best_neighbor, min_cost, all_successors

def hill_climbing(initial_state):
    current_state = tuple(initial_state)
    current_cost = calculate_cost(current_state)

    step = 0
    print("\nUSN:1BM24CS104 NAME:GEETHA T")
    print("\n=============================================")
    print("         HILL CLIMBING SEARCH FLOW           ")
    print("=============================================")
    print(f"Step {step} (Initial State): {current_state} | Cost: {current_cost}")

    # Check if the initial state itself is already a solution
    if current_cost == 0:
        print("\n=============================================")
        print(f"🎉 SUCCESS! The initial state is already a valid solution: {current_state}")
        print("=============================================")
        return current_state

    while True:
        # Generate neighbors of the current state
        neighbor_state, neighbor_cost, all_successors = get_best_successor(current_state)

        print(f"\n--- Successors generated from {current_state} ---")
        for succ, cost, swap in all_successors:
            marker = " <- [Best Successor]" if succ == neighbor_state else ""
            print(f"  Swap columns {swap} -> State: {succ} | Cost: {cost}{marker}")

        # Termination check: if neighbor cannot improve the current cost, stop.
        if neighbor_cost >= current_cost:
            print("\n=============================================")
            print("🛑 Stopping Condition Reached (neighbor.COST >= current.COST)")
            print(f"⚠️ STUCK at Local Minimum / Plateau: {current_state}")
            print("=============================================")
            return current_state

        # Move to the better neighbor
        step += 1
        current_state = neighbor_state
        current_cost = neighbor_cost
        print(f"\n➔ Step {step}: Moving to lower cost state {current_state} | Cost: {current_cost}")

        # CRITICAL OPTIMIZATION: Stop immediately if a perfect solution is found
        if current_cost == 0:
            print("\n=============================================")
            print(f"🎉 SUCCESS! Found a valid perfect solution: {current_state}")
            print("=============================================")
            return current_state

if __name__ == "__main__":
    print("Enter the initial positions for the 4 queens (a permutation of 0, 1, 2, 3).")
    print("Example: 1 3 2 0")

    while True:
        try:
            user_input = input("Enter 4 space-separated numbers: ").strip().split()
            initial_board = [int(x) for x in user_input]
            if len(initial_board) != 4 or set(initial_board) != {0, 1, 2, 3}:
                print("Invalid configuration! Must be a permutation of 0, 1, 2, 3.")
                continue
            break
        except ValueError:
            print("Invalid input. Please enter numbers only.")

    hill_climbing(initial_board)
```
