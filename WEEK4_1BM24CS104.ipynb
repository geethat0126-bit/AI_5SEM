```python
import math

def generate_sa_table():
    print("=== Simulated Annealing Table Generator ===")

    # 1. Take Temperature inputs from the user
    t_input = input("Enter temperatures separated by commas (e.g., 100, 50, 10): ")
    temperatures = [float(t.strip()) for t in t_input.split(",") if t.strip()]

    # 2. Take Cost inputs from the user
    print("\nEnter pairs of (current_cost, next_cost) separated by a comma.")
    print("Type 'done' when you are finished entering rows.")

    rows = []
    while True:
        user_in = input(f"Row {len(rows)+1} [current, next] (or 'done'): ").strip()
        if user_in.lower() == 'done':
            break
        try:
            c_cost, n_cost = map(float, user_in.split(","))
            rows.append((c_cost, n_cost))
        except ValueError:
            print("Invalid input! Please enter two numbers separated by a comma.")

    if not rows or not temperatures:
        print("Missing inputs. Cannot generate table.")
        return

    # 3. Generate Table Headers
    headers = ["current.cost", "next.cost", "ΔE"] + [f"p(T={int(t) if t.is_integer() else t})" for t in temperatures]
    print("\n### Generated Table:\n")
    print("| " + " | ".join(headers) + " |")
    print("| " + " | ".join(["---"] * len(headers)) + " |")

    # 4. Calculate and Print Data Rows
    for c_cost, n_cost in rows:
        delta_e = c_cost - n_cost
        prob_cols = []

        for t in temperatures:
            if delta_e > 0:
                prob_str = "1"
            else:
                prob = math.exp(delta_e / t)
                # Format to scientific notation if numbers are extremely small
                prob_str = f"{prob:.1e}" if prob < 0.001 else f"{prob:.2f}"
            prob_cols.append(prob_str)

        # Clean up floating points to integers if they are whole numbers
        c_str = f"{int(c_cost)}" if c_cost.is_integer() else f"{c_cost}"
        n_str = f"{int(n_cost)}" if n_cost.is_integer() else f"{n_cost}"
        d_str = f"{int(delta_e)}" if delta_e.is_integer() else f"{delta_e}"

        row_data = [c_str, n_str, d_str] + prob_cols
        print("| " + " | ".join(row_data) + " |")

if __name__ == "__main__":
    generate_sa_table()
```
