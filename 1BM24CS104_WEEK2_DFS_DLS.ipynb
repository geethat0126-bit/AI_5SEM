```python
class EightPuzzleDFS:
    def __init__(self, start_state, goal_state):
        self.start_state = tuple(start_state)
        self.goal_state = tuple(goal_state)
        self.moves = {'Up': -3, 'Down': 3, 'Left': -1, 'Right': 1}

    def get_neighbors(self, state):
        neighbors = []
        zero_idx = state.index(0)
        row, col = divmod(zero_idx, 3)

        for direction, offset in self.moves.items():
            new_idx = zero_idx + offset
            new_row, new_col = divmod(new_idx, 3)

            if 0 <= new_row < 3 and 0 <= new_col < 3:
                if direction in ['Left', 'Right'] and row != new_row:
                    continue
                if direction in ['Up', 'Down'] and col != new_col:
                    continue

                neighbor_list = list(state)
                neighbor_list[zero_idx], neighbor_list[new_idx] = neighbor_list[new_idx], neighbor_list[zero_idx]
                neighbors.append((tuple(neighbor_list), direction))

        return neighbors

    def solve(self, max_depth=30):
        stack = [(self.start_state, [], 0)]
        visited = set()

        while stack:
            state, path, depth = stack.pop()

            if state == self.goal_state:
                return path

            if depth >= max_depth:
                continue

            if state not in visited:
                visited.add(state)
                for neighbor, direction in reversed(self.get_neighbors(state)):
                    if neighbor not in visited:
                        stack.append((neighbor, path + [direction], depth + 1))
        return None

def print_board(state):
    for i in range(0, 9, 3):
        print(f"{state[i]} {state[i+1]} {state[i+2]}")
    print()

def get_user_input(state_name):
    """Prompts the user for a valid 8-puzzle board configuration."""
    print(f"\n--- Enter numbers for the {state_name} ---")
    print("Rules: Use numbers 0-8 (0 represents the empty space).")
    print("Format option A: Type all 9 numbers separated by spaces on one line (e.g., 1 2 3 4 5 6 7 8 0)")
    print("Format option B: Type 3 numbers per line for 3 lines.")

    while True:
        try:
            inputs = []
            # Read first line
            line = input("Enter values: ").strip()
            inputs.extend(line.split())

            # If they entered fewer than 9 numbers, keep reading lines until we hit 9
            while len(inputs) < 9:
                next_line = input().strip()
                if not next_line:  # Skip accidental empty hits
                    continue
                inputs.extend(next_line.split())

            # Trim if they accidentally added more
            if len(inputs) > 9:
                inputs = inputs[:9]

            # Convert to integers
            state = [int(x) for x in inputs]

            # Validate input matches exactly 0-8 elements
            if sorted(state) != list(range(9)):
                print("❌ Invalid input! The state must contain exactly the numbers 0 through 8 with no duplicates. Try again.")
                continue

            return state

        except ValueError:
            print("❌ Invalid characters detected! Please use integers only. Try again.")

# --- Execution ---
if __name__ == "__main__":
    print("🧩 8-Puzzle DFS Solver 🧩")
    print("NAME:GEETHA T  USN:1BM24CS104")

    # Capture user configuration
    initial_state = get_user_input("INITIAL STATE")
    goal_state = get_user_input("GOAL STATE")

    # Confirm configurations back to the user
    print("\nInitial Board:")
    print_board(initial_state)
    print("Goal Board Target:")
    print_board(goal_state)

    # Execute solver
    solver = EightPuzzleDFS(initial_state, goal_state)
    print("Searching for solution via DFS...")

    # 30 is a safe depth budget for small deviations, raise if testing complex goal shifts
    solution_path = solver.solve(max_depth=30)

    if solution_path:
        print(f"🎉 Solution found in {len(solution_path)} moves!")
        print("Sequence of moves:", " -> ".join(solution_path))
    else:
        print("❌ No solution found within the specified maximum depth limit.")
```

```python
class EightPuzzleIDDFS:
    def __init__(self, start_state, goal_state):
        self.start_state = tuple(start_state)
        self.goal_state = tuple(goal_state)
        self.moves = {'Up': -3, 'Down': 3, 'Left': -1, 'Right': 1}

    def get_neighbors(self, state):
        """Generates valid adjacent states."""
        neighbors = []
        zero_idx = state.index(0)
        row, col = divmod(zero_idx, 3)

        for direction, offset in self.moves.items():
            new_idx = zero_idx + offset
            new_row, new_col = divmod(new_idx, 3)

            if 0 <= new_row < 3 and 0 <= new_col < 3:
                if direction in ['Left', 'Right'] and row != new_row:
                    continue
                if direction in ['Up', 'Down'] and col != new_col:
                    continue

                neighbor_list = list(state)
                neighbor_list[zero_idx], neighbor_list[new_idx] = neighbor_list[new_idx], neighbor_list[zero_idx]
                neighbors.append((tuple(neighbor_list), direction))

        return neighbors

    def depth_limited_search(self, limit):
        """Performs a depth-limited search using a LIFO stack."""
        stack = [(self.start_state, [], 0)]

        while stack:
            state, path, depth = stack.pop()

            if state == self.goal_state:
                return path

            if depth >= limit:
                continue

            for neighbor, direction in reversed(self.get_neighbors(state)):
                # Simple cycle check along the active branch path
                if neighbor not in path:
                    stack.append((neighbor, path + [direction], depth + 1))

        return None

    def solve(self, max_depth_limit):
        """Iteratively increases the depth limit up to the user-specified maximum."""
        for depth in range(max_depth_limit + 1):
            print(f"Searching at depth limit: {depth}...")
            result = self.depth_limited_search(depth)
            if result is not None:
                return result
        return None

def print_board(state):
    for i in range(0, 9, 3):
        print(f"{state[i]} {state[i+1]} {state[i+2]}")
    print()

def get_user_input(state_name):
    print(f"\n--- Enter numbers for the {state_name} ---")
    print("Rules: Use numbers 0-8 (0 is the empty space).")
    print("Example layout format: 1 2 3 4 5 6 7 8 0")

    while True:
        try:
            inputs = []
            line = input("Enter values: ").strip()
            inputs.extend(line.split())

            while len(inputs) < 9:
                next_line = input().strip()
                if not next_line:
                    continue
                inputs.extend(next_line.split())

            if len(inputs) > 9:
                inputs = inputs[:9]

            state = [int(x) for x in inputs]
            if sorted(state) != list(range(9)):
                print("❌ Invalid input! Must contain exactly digits 0-8 with no duplicates.")
                continue

            return state
        except ValueError:
            print("❌ Invalid characters! Please enter integers only.")

def get_depth_limit():
    """Prompts the user for a valid, positive integer depth limit."""
    while True:
        try:
            limit = int(input("\nEnter the maximum depth limit to search within: ").strip())
            if limit < 0:
                print("❌ The depth limit must be 0 or a positive integer.")
                continue
            return limit
        except ValueError:
            print("❌ Invalid input! Please enter a valid integer for the depth limit.")

# --- Execution ---
if __name__ == "__main__":
    print("🧩 8-Puzzle IDDFS Solver 🧩")
    print("NAME:GEETHA T   USN:1BM24CS104")

    # 1. Get States from User
    initial_state = get_user_input("INITIAL STATE")
    goal_state = get_user_input("GOAL STATE")

    # 2. Get Depth Limit from User
    max_depth = get_depth_limit()

    print("\nInitial Board:")
    print_board(initial_state)
    print("Goal Board Target:")
    print_board(goal_state)
    print(f"Maximum Depth Budget set to: {max_depth}\n")

    # 3. Process Solution
    solver = EightPuzzleIDDFS(initial_state, goal_state)
    solution_path = solver.solve(max_depth)

    if solution_path is not None:
        print(f"\n🎉 Success! Optimal solution found in {len(solution_path)} moves within your depth limit.")
        print("Sequence of moves:", " -> ".join(solution_path) if solution_path else "Already at goal state!")
    else:
        print(f"\n❌ Failure! No solution could be found within your depth limit of {max_depth} moves.")
```
