```python
import heapq

class PuzzleState:
    def __init__(self, board, goal_board, g=0, parent=None, move="Initial"):
        self.board = board          # 1D tuple of length 9
        self.goal_board = goal_board  # 1D tuple of length 9
        self.g = g                  # Cost from start to current node
        self.parent = parent        # Parent state node
        self.move = move            # Move taken to reach this state (Left, Up, Right, Down)
        self.h = self.calculate_misplaced_tiles()
        self.f = self.g + self.h

    def calculate_misplaced_tiles(self):
        """Calculates h value using Misplaced Tiles heuristic."""
        count = 0
        for i in range(9):
            if self.board[i] != 0 and self.board[i] != self.goal_board[i]:
                count += 1
        return count

    def get_neighbors(self):
        """Generates valid next states along with their move direction label."""
        neighbors = []
        blank_idx = self.board.index(0)
        row, col = blank_idx // 3, blank_idx % 3

        # Left, Up, Right, Down
        moves = [
            (0, -1, "Left"),
            (-1, 0, "Up"),
            (0, 1, "Right"),
            (1, 0, "Down")
        ]

        for dr, dc, move_label in moves:
            new_row, new_col = row + dr, col + dc
            if 0 <= new_row < 3 and 0 <= new_col < 3:
                new_blank_idx = new_row * 3 + new_col

                new_board = list(self.board)
                new_board[blank_idx], new_board[new_blank_idx] = new_board[new_blank_idx], new_board[blank_idx]

                neighbors.append(PuzzleState(tuple(new_board), self.goal_board, self.g + 1, self, move_label))
        return neighbors

    def __lt__(self, other):
        if self.f == other.f:
            return self.g > other.g
        return self.f < other.f


def parse_input_state(prompt_text):
    """Helper function to parse user string into a 1D tuple format."""
    print(prompt_text)
    print("Enter 3 numbers per row separated by spaces (use 0 for blank space):")
    board_list = []
    for i in range(3):
        while True:
            try:
                row = list(map(int, input(f"Row {i+1}: ").strip().split()))
                if len(row) != 3:
                    print("Error: Please enter exactly 3 numbers.")
                    continue
                board_list.extend(row)
                break
            except ValueError:
                print("Error: Invalid numeric input. Try again.")
    return tuple(board_list)


def print_formatted_node(node):
    """Prints a node state block with board, costs, and depth levels."""
    print(f"--- Node State (via Move: {node.move}) ---")
    board = node.board
    for i in range(0, 9, 3):
        print(f"  {board[i]} {board[i+1]} {board[i+2]}")
    print(f"  g = {node.g}")
    print(f"  h = {node.h}")
    print(f"  f = {node.g} + {node.h} = {node.f}\n")


def a_star_search(initial_board, goal_board):
    """Executes search and logs details of the evaluated state pathways."""
    start_node = PuzzleState(initial_board, goal_board, g=0)
    frontier = []
    heapq.heappush(frontier, start_node)

    explored = set()
    step_counter = 0
    print("USN:1BM24CS104 NAME:GEETHA T")
    print("\n=============================================")
    print("           STARTING SEARCH FLOW             ")
    print("=============================================\n")

    while frontier:
        current_node = heapq.heappop(frontier)
        step_counter += 1

        print(f"[Evaluation Step #{step_counter}]")
        print_formatted_node(current_node)

        # Stop immediately when the goal is reached without printing any optimal path trajectory below
        if current_node.board == goal_board:
            print("✔ GOAL STATE REACHED!")
            return current_node

        explored.add(current_node.board)

        for neighbor in current_node.get_neighbors():
            if neighbor.board in explored:
                continue
            heapq.heappush(frontier, neighbor)

    print("❌ Frontier empty. No solution path exists.")
    return None


if __name__ == "__main__":
    initial_state = parse_input_state("\n--- Define Initial Start State ---")
    final_state = parse_input_state("\n--- Define Target Final Goal State ---")

    # This will run the search flow and terminate as soon as the goal is found
    a_star_search(initial_state, final_state)
```
