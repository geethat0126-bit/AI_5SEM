```python
import random
import time

def print_board(board):
    """Prints the current state of the 3x3 game board."""
    print("\n")
    print(f" {board[0]} | {board[1]} | {board[2]} ")
    print("---|---|---")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("---|---|---")
    print(f" {board[6]} | {board[7]} | {board[8]} ")
    print("\n")

def check_winner(board, player):
    """Checks if the given player has won the game."""
    # The 8 possible winning index combinations on a 3x3 grid
    win_conditions = [
        [0, 1, 2], [3, 4, 5], [6, 7, 8],  # Rows
        [0, 3, 6], [1, 4, 7], [2, 5, 8],  # Columns
        [0, 4, 8], [2, 4, 6]              # Diagonals
    ]
    for condition in win_conditions:
        if board[condition[0]] == board[condition[1]] == board[condition[2]] == player:
            return True
    return False

def check_draw(board):
    """Checks if the board is completely full, resulting in a draw."""
    return all(space in ['X', 'O'] for space in board)

def get_computer_move(board):
    """Finds all available spaces and picks one at random."""
    available_moves = [i for i, space in enumerate(board) if space not in ['X', 'O']]
    return random.choice(available_moves)

def play_game():
    """Main game loop managing turns between Human and Computer."""
    # Initialize the board with numbers 1-9 representing the spaces
    board = [str(i) for i in range(1, 10)]
    human = 'X'
    computer = 'O'
    current_turn = 'Human'  # Human starts first
    game_over = False

    print("Welcome to Tic-Tac-Toe: Geetha T USN:1BM24CS104 vs. Computer!")
    print("You are 'X' and the Computer is 'O'.")
    print("Enter a number from 1-9 to place your mark.")

    while not game_over:
        print_board(board)

        if current_turn == 'Human':
            # --- HUMAN TURN ---
            try:
                choice = int(input("Your turn! Choose a spot (1-9): ")) - 1
            except ValueError:
                print("Invalid input! Please enter a number between 1 and 9.")
                continue

            if choice < 0 or choice > 8:
                print("Out of range! Please pick a spot between 1 and 9.")
                continue
            elif board[choice] in ['X', 'O']:
                print("That spot is already taken! Pick another one.")
                continue

            # Apply Human Move
            board[choice] = human

            # Check game status
            if check_winner(board, human):
                print_board(board)
                print("🎉 Congratulations! You beat the computer! 🎉")
                game_over = True
            elif check_draw(board):
                print_board(board)
                print("🤝 It's a draw!")
                game_over = True
            else:
                current_turn = 'Computer'

        else:
            # --- COMPUTER TURN ---
            print("Computer is thinking...")
            time.sleep(1) # Adds a small, realistic delay to the computer's turn

            choice = get_computer_move(board)
            board[choice] = computer
            print(f"Computer picked spot {choice + 1}.")

            # Check game status
            if check_winner(board, computer):
                print_board(board)
                print("🤖 The computer wins! Better luck next time.")
                game_over = True
            elif check_draw(board):
                print_board(board)
                print("🤝 It's a draw!")
                game_over = True
            else:
                current_turn = 'Human'

if __name__ == "__main__":
    play_game()
```
