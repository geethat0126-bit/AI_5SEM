```python
import random
import time

def print_board(board):
    """Prints the current state of the 3x3 game board."""
    print("\n")
    print(f" {board[0]} | {board[1]} | {board[2]} ")
    print("---|---|---")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("---|---|---")
    print(f" {board[6]} | {board[7]} | {board[8]} ")
    print("\n")

def check_winner(board, player):
    """Checks if the given player has won the game."""
    # The 8 possible winning index combinations on a 3x3 grid
    win_conditions = [
        [0, 1, 2], [3, 4, 5], [6, 7, 8],  # Rows
        [0, 3, 6], [1, 4, 7], [2, 5, 8],  # Columns
        [0, 4, 8], [2, 4, 6]              # Diagonals
    ]
    for condition in win_conditions:
        if board[condition[0]] == board[condition[1]] == board[condition[2]] == player:
            return True
    return False

def check_draw(board):
    """Checks if the board is completely full, resulting in a draw."""
    return all(space in ['X', 'O'] for space in board)

def get_computer_move(board):
    """Finds all available spaces and picks one at random."""
    available_moves = [i for i, space in enumerate(board) if space not in ['X', 'O']]
    return random.choice(available_moves)

def play_game():
    """Main game loop managing turns between Human and Computer."""
    # Initialize the board with numbers 1-9 representing the spaces
    board = [str(i) for i in range(1, 10)]
    human = 'X'
    computer = 'O'
    current_turn = 'Human'  # Human starts first
    game_over = False

    print("Welcome to Tic-Tac-Toe: Geetha T USN:1BM24CS104 vs. Computer!")
    print("You are 'X' and the Computer is 'O'.")
    print("Enter a number from 1-9 to place your mark.")

    while not game_over:
        print_board(board)

        if current_turn == 'Human':
            # --- HUMAN TURN ---
            try:
                choice = int(input("Your turn! Choose a spot (1-9): ")) - 1
            except ValueError:
                print("Invalid input! Please enter a number between 1 and 9.")
                continue

            if choice < 0 or choice > 8:
                print("Out of range! Please pick a spot between 1 and 9.")
                continue
            elif board[choice] in ['X', 'O']:
                print("That spot is already taken! Pick another one.")
                continue

            # Apply Human Move
            board[choice] = human

            # Check game status
            if check_winner(board, human):
                print_board(board)
                print("🎉 Congratulations! You beat the computer! 🎉")
                game_over = True
            elif check_draw(board):
                print_board(board)
                print("🤝 It's a draw!")
                game_over = True
            else:
                current_turn = 'Computer'

        else:
            # --- COMPUTER TURN ---
            print("Computer is thinking...")
            time.sleep(1) # Adds a small, realistic delay to the computer's turn

            choice = get_computer_move(board)
            board[choice] = computer
            print(f"Computer picked spot {choice + 1}.")

            # Check game status
            if check_winner(board, computer):
                print_board(board)
                print("🤖 The computer wins! Better luck next time.")
                game_over = True
            elif check_draw(board):
                print_board(board)
                print("🤝 It's a draw!")
                game_over = True
            else:
                current_turn = 'Human'

if __name__ == "__main__":
    play_game()
```

```python
import time

class VacuumEnvironment:
    """
    Simulates a 2-room environment shown as a box layout.
    """
    def __init__(self, room_a_status, room_b_status, starting_location):
        self.status = {
            'A': room_a_status,
            'B': room_b_status
        }
        self.agent_location = starting_location

    def display_box_simulation(self, step, operating_cost):
        """Renders the environment inside a clear visual box layout with live cost tracking."""
        print(f"\n================ STEP {step} ================")
        print(f" Total Cost: {operating_cost} units")

        # Determine labels for vacuum presence
        vac_a = " [🤖 VACUUM] " if self.agent_location == 'A' else "            "
        vac_b = " [🤖 VACUUM] " if self.agent_location == 'B' else "            "

        # Pad strings for perfect box formatting
        status_a = self.status['A'].center(12)
        status_b = self.status['B'].center(12)

        # Draw the simulation grid box
        print("┌───────────────────────────┬───────────────────────────┐")
        print("│          ROOM A           │          ROOM B           │")
        print("├───────────────────────────┼───────────────────────────┤")
        print(f"│       Status: {status_a}│       Status: {status_b}│")
        print("│                           │                           │")
        print(f"│      {vac_a}       │      {vac_b}       │")
        print("└───────────────────────────┴───────────────────────────┘")


class SimpleReflexVacuumAgent:
    """
    An AI Simple Reflex Agent operating on condition-action rules with cost calculations.
    """
    def __init__(self):
        self.cost = 0

    def perceive_and_act(self, env):
        location = env.agent_location
        status = env.status[location]

        print(f"-> Percept: Location {location} is {status}")
        time.sleep(1.0) # Pause so you can track the visual updates

        if status == 'Dirty':
            print(f"💥 Action: SUCK (Extracting dust from Room {location}...)")
            env.status[location] = 'Clean'
            self.cost += 1     # Cost for cleaning is 1
        else:
            if location == 'A':
                print("➡️ Action: MOVE RIGHT (Transitioning to Room B...)")
                env.agent_location = 'B'
            else:
                print("⬅️ Action: MOVE LEFT (Transitioning to Room A...)")
                env.agent_location = 'A'
            self.cost += 0     # Cost for moving is 0
        time.sleep(1.0)

    def get_cost(self):
        return self.cost


# Helper function to get clean/dirty input validation
def get_room_status(room_name):
    while True:
        status = input(f"Enter status for Room {room_name} (Clean/Dirty): ").strip().capitalize()
        if status in ['Clean', 'Dirty']:
            return status
        print("Invalid input! Please type either 'Clean' or 'Dirty'.")

# Helper function to get location validation
def get_agent_location():
    while True:
        location = input("Enter vacuum starting location (A/B): ").strip().upper()
        if location in ['A', 'B']:
            return location
        print("Invalid input! Please type either 'A' or 'B'.")


# Simulation Execution
if __name__ == "__main__":
    print("--- Artificial Intelligence Vacuum Cleaner Box Simulation ---")
    print("Name:Geetha T USN:1BM24CS104c")
    print("Please set up the initial environment:\n")

    # Prompting manual user inputs
    status_a = get_room_status('A')
    status_b = get_room_status('B')
    start_loc = get_agent_location()

    # Initialize the custom environment and agent
    environment = VacuumEnvironment(status_a, status_b, start_loc)
    agent = SimpleReflexVacuumAgent()

    print("\n--- Initializing Box Simulation Grid ---")

    # Run the simulation for 4 steps to watch the boxes change states dynamically
    for step in range(1, 5):
        environment.display_box_simulation(step, agent.get_cost())
        agent.perceive_and_act(environment)

    print("\n================ SIMULATION ENDED ================")
    environment.display_box_simulation("FINAL", agent.get_cost())
    print(f"Total Resource Cost Accrued:   {agent.get_cost()} units")
```
