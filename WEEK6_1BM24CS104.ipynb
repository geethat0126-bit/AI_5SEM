```python
import itertools

def generate_propositional_table():
    # Define the headers matching the image exactly
    headers = ["A", "B", "C", "A v C", "B v ~C", "KB", "α"]
    print("NAME:GEETHA T USN:1BM24CS104")
    # Print Markdown table headers
    print("| " + " | ".join(headers) + " |")
    print("|" + "|".join(["---"] * len(headers)) + "|")

    # Generate all combinations of False and True for A, B, C (000 to 111)
    for combo in itertools.product([False, True], repeat=3):
        A, B, C = combo

        # Evaluate clauses and expressions
        A_v_C = A or C
        B_v_notC = B or (not C)
        KB = A_v_C and B_v_notC
        alpha = A or B

        # Helper function to convert Python booleans to lowercase string format
        def format_bool(val):
            return "true" if val else "false"

        # Build and print the row
        row = [
            format_bool(A),
            format_bool(B),
            format_bool(C),
            format_bool(A_v_C),
            format_bool(B_v_notC),
            format_bool(KB),
            format_bool(alpha)
        ]
        print("| " + " | ".join(row) + " |")

if __name__ == "__main__":
    generate_propositional_table()
```
